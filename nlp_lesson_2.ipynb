```python
import re

text = "HEY BRO!!! Visit https://example.com 😊 WIN 100% FREE!!!"

# Convert to lowercase
cleaned_text = text.lower()

# Remove URLs
cleaned_text = re.sub(r"https?://\S+|www\.\S+", "", cleaned_text)

# Remove punctuation
cleaned_text = re.sub(r"[^\w\s]", "", cleaned_text)

# Remove extra spaces
cleaned_text = re.sub(r"\s+", " ", cleaned_text).strip()

print("Original:", text)
print("Cleaned :", cleaned_text)
```

```python
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

# Preserve "not" and "no" because they can change meaning
stop_words.discard("not")
stop_words.discard("no")

def preprocess_text(text):
    # 1. Normalize case
    text = text.lower()

    # 2. Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # 3. Keep letters and spaces for this example
    text = re.sub(r"[^a-z\s]", " ", text)

    # 4. Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # 5. Tokenize
    tokens = word_tokenize(text)

    # 6. Remove stop words
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # 7. Lemmatize as nouns by default
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    return tokens


message = "CONGRATULATIONS!!! You have won a free prize. Click https://example.com"

result = preprocess_text(message)

print("Processed tokens:", result)
print("Processed text:", " ".join(result))
```

```python

# ============================================================
# MISSION: AI ENGINEER 2027
# NLP LESSON 3: BAG OF WORDS, N-GRAMS AND TF-IDF
# ============================================================

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

# STEP 1: Create a small text dataset
documents = [
    "Python is useful for AI",
    "Python is useful for data science",
    "AI helps build intelligent systems",
    "I love learning Python",
    "I love learning AI"
]

print("=" * 60)
print("1. ORIGINAL TEXT DATASET")
print("=" * 60)

for i, text in enumerate(documents, start=1):
    print(f"Document {i}: {text}")


# STEP 2: BAG OF WORDS (WORD COUNTS)
print("\n" + "=" * 60)
print("2. BAG OF WORDS")
print("=" * 60)

bow_vectorizer = CountVectorizer()
bow_matrix = bow_vectorizer.fit_transform(documents)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow_vectorizer.get_feature_names_out()
)

print("\nVocabulary:")
print(bow_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_df)


# STEP 3: N-GRAMS (UNIGRAMS + BIGRAMS)
print("\n" + "=" * 60)
print("3. N-GRAMS: UNIGRAMS AND BIGRAMS")
print("=" * 60)

ngram_vectorizer = CountVectorizer(ngram_range=(1, 2))
ngram_matrix = ngram_vectorizer.fit_transform(documents)

ngram_df = pd.DataFrame(
    ngram_matrix.toarray(),
    columns=ngram_vectorizer.get_feature_names_out()
)

print("\nUnigrams and Bigrams:")
print(ngram_vectorizer.get_feature_names_out())

print("\nN-gram Matrix:")
print(ngram_df)


# STEP 4: TF-IDF (WORD IMPORTANCE WEIGHTS)
print("\n" + "=" * 60)
print("4. TF-IDF")
print("=" * 60)

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("\nTF-IDF Matrix (rounded to 3 decimals):")
print(tfidf_df.round(3))


# STEP 5: COMPARE BOW COUNTS WITH TF-IDF SCORES
print("\n" + "=" * 60)
print("5. COMPARISON")
print("=" * 60)

print("\nFirst document:")
print(documents[0])

print("\nBag of Words values:")
print(bow_df.iloc[0][bow_df.iloc[0] > 0])

print("\nTF-IDF values:")
print(tfidf_df.iloc[0][tfidf_df.iloc[0] > 0].round(3))


# STEP 6: TEST WITH A NEW MESSAGE
print("\n" + "=" * 60)
print("6. TRANSFORM A NEW MESSAGE")
print("=" * 60)

new_message = ["I love Python AI"]

new_bow = bow_vectorizer.transform(new_message)
new_tfidf = tfidf_vectorizer.transform(new_message)

print("\nNew message:", new_message[0])

print("\nNew message - Bag of Words:")
print(pd.DataFrame(
    new_bow.toarray(),
    columns=bow_vectorizer.get_feature_names_out()
))

print("\nNew message - TF-IDF:")
print(pd.DataFrame(
    new_tfidf.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
).round(3))

print("\n" + "=" * 60)
print("LESSON 3 COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
